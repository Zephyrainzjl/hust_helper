## 非破坏式增删改（写入用户 overlay）


```python
new_entry = eater.add(
    name="我的新发现",
    chapter_title="用户扩展",
    section_title="华科周边",
    description="可以自由添加字段和图片。",
    tags=["用户扩展"],
)
new_entry.id
```

```python
eater.update(new_entry.id, description="已通过 overlay 更新")
eater.delete(new_entry.id)
eater.restore(new_entry.id)
```
